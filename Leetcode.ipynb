## 242. Valid Anagram

```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        s_dict = {}
        t_dict = {}
        if len(s) != len(t):
            return False

        for i in range(len(s)):
            s_dict[s[i]] = 1+s_dict.get(s[i], 0)
            t_dict[t[i]] = 1+t_dict.get(t[i], 0)

        for j in s_dict:
            if s_dict[j] != t_dict.get(j, 0):
                return False

        return True
```
